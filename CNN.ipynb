## Cấu hình và huấn luyện


```python
from __future__ import annotations

import json
import random
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf
from IPython.display import display
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import ImageDataGenerator


MODEL_NAME = "CNN"
PREPROCESSED_ROOT = Path("/content/dataset_classification")
RESULTS_ROOT = PREPROCESSED_ROOT.parent / "training_results"
RESULTS_DIR = RESULTS_ROOT / MODEL_NAME

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 100
RUNS = 1

# De chay dung 100 epoch, mac dinh tat EarlyStopping.
USE_EARLY_STOPPING = False
EARLY_STOPPING_PATIENCE = 10

LEARNING_RATE = 1e-3
SEED = 42
EXPECTED_CLASS_ORDER = ("bad", "good", "shell")
IMAGE_SUFFIXES = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
OUTPUT_PATTERNS_TO_REMOVE = [
    f"{MODEL_NAME}_history_run*.csv",
    f"{MODEL_NAME}_classification_report_*.csv",
    f"{MODEL_NAME}_summary_*.csv",
    f"{MODEL_NAME}_split_metrics.csv",
    f"{MODEL_NAME}_test_predictions.csv",
    f"{MODEL_NAME}_run_summary.csv",
    f"{MODEL_NAME}_*.png",
]


def configure_environment() -> None:
    random.seed(SEED)
    np.random.seed(SEED)
    tf.keras.utils.set_random_seed(SEED)

    gpus = tf.config.list_physical_devices("GPU")
    if gpus:
        try:
            for gpu in gpus:
                tf.config.experimental.set_memory_growth(gpu, True)
        except RuntimeError as exc:
            print("Khong the bat memory growth vi GPU da duoc khoi tao:", exc)
        print("Dang uu tien chay bang GPU:", gpus)
    else:
        print("Khong tim thay GPU. TensorFlow se chay bang CPU. Neu dung Colab, vao Runtime > Change runtime type > GPU.")

    RESULTS_DIR.mkdir(parents=True, exist_ok=True)
    print("Thu muc luu ket qua:", RESULTS_DIR)


def cleanup_old_outputs(workdir: Path) -> None:
    workdir.mkdir(parents=True, exist_ok=True)
    for pattern in OUTPUT_PATTERNS_TO_REMOVE:
        for output_path in workdir.glob(pattern):
            if output_path.is_file():
                output_path.unlink()


def save_config(class_names: list[str]) -> None:
    config = {
        "model_name": MODEL_NAME,
        "dataset_path": str(PREPROCESSED_ROOT),
        "results_dir": str(RESULTS_DIR),
        "image_size": IMAGE_SIZE,
        "batch_size": BATCH_SIZE,
        "epochs": EPOCHS,
        "runs": RUNS,
        "use_early_stopping": USE_EARLY_STOPPING,
        "learning_rate": LEARNING_RATE,
        "class_names": class_names,
    }
    with open(RESULTS_DIR / f"{MODEL_NAME}_config.json", "w", encoding="utf-8") as file:
        json.dump(config, file, ensure_ascii=False, indent=2)


def list_class_names_from_indices(class_indices: dict[str, int]) -> list[str]:
    return [
        class_name
        for class_name, _ in sorted(class_indices.items(), key=lambda item: item[1])
    ]


def check_class_order(class_names: list[str]) -> None:
    if tuple(class_names) != EXPECTED_CLASS_ORDER:
        print("Canh bao: thu tu lop hien tai la", class_names)
        print("Thu tu mong muon:", EXPECTED_CLASS_ORDER)
        print("Neu ten folder la bad, good, shell thi Keras thuong tu sap xep dung theo alphabet.")


def create_generators(data_root: Path):
    datagen = ImageDataGenerator(rescale=1.0 / 255.0)

    train_gen = datagen.flow_from_directory(
        data_root / "train",
        target_size=IMAGE_SIZE,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        shuffle=True,
        seed=SEED,
    )

    train_eval_gen = datagen.flow_from_directory(
        data_root / "train",
        target_size=IMAGE_SIZE,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        shuffle=False,
    )

    val_gen = datagen.flow_from_directory(
        data_root / "val",
        target_size=IMAGE_SIZE,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        shuffle=False,
    )

    test_dir = data_root / "test"
    test_gen = None
    if test_dir.exists():
        test_gen = datagen.flow_from_directory(
            test_dir,
            target_size=IMAGE_SIZE,
            batch_size=BATCH_SIZE,
            class_mode="categorical",
            shuffle=False,
        )

    class_names = list_class_names_from_indices(train_gen.class_indices)
    print("Class mapping:", train_gen.class_indices)
    pd.DataFrame(
        [{"class_name": name, "class_index": index} for name, index in train_gen.class_indices.items()]
    ).sort_values("class_index").to_csv(RESULTS_DIR / f"{MODEL_NAME}_class_indices.csv", index=False)
    check_class_order(class_names)
    return train_gen, train_eval_gen, val_gen, test_gen


def build_class_weight(class_indices: dict[str, int], labels: np.ndarray) -> dict[int, float]:
    counts = np.bincount(labels.astype(int), minlength=len(class_indices))
    total = int(counts.sum())
    num_classes = len(class_indices)
    class_weight: dict[int, float] = {}

    class_count_by_name = {
        class_name: int(counts[index])
        for class_name, index in sorted(class_indices.items(), key=lambda item: item[1])
    }
    print("Class counts:", class_count_by_name)
    pd.DataFrame(
        [{"class_name": name, "count": count} for name, count in class_count_by_name.items()]
    ).to_csv(RESULTS_DIR / f"{MODEL_NAME}_class_counts.csv", index=False)

    for class_name, index in class_indices.items():
        class_count = int(counts[index])
        if class_count == 0:
            raise ValueError(f"Class '{class_name}' has no samples in the training set.")
        class_weight[index] = total / (num_classes * class_count)

    print("Class weight:", class_weight)
    pd.DataFrame(
        [{"class_index": index, "class_weight": weight} for index, weight in class_weight.items()]
    ).sort_values("class_index").to_csv(RESULTS_DIR / f"{MODEL_NAME}_class_weight.csv", index=False)
    return class_weight


def show_batch_samples(train_gen) -> None:
    images, labels = next(train_gen)
    class_names = list_class_names_from_indices(train_gen.class_indices)

    plt.figure(figsize=(12, 4))
    for i in range(min(6, len(images))):
        label_index = int(np.argmax(labels[i]))
        plt.subplot(2, 3, i + 1)
        plt.imshow(images[i])
        plt.title(class_names[label_index])
        plt.axis("off")

    plt.suptitle("CNN input preview (224x224, normalized by /255.0)")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_preprocessed_samples.png", dpi=150)
    plt.show()
    plt.close()
    train_gen.reset()


def create_model(num_classes: int) -> tf.keras.Model:
    model = models.Sequential([
        layers.Input(shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3)),

        layers.Conv2D(32, 3, activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(64, 3, activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(128, 3, activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(256, 3, activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.4),
        layers.Dense(128, activation="relu"),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation="softmax"),
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE),
        loss="categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model


def plot_history(history_df: pd.DataFrame) -> None:
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history_df["epoch"], history_df["accuracy"], label="Train Accuracy")
    plt.plot(history_df["epoch"], history_df["val_accuracy"], label="Val Accuracy")
    plt.legend()
    plt.title("Accuracy by Epoch")
    plt.xlabel("Epoch")

    plt.subplot(1, 2, 2)
    plt.plot(history_df["epoch"], history_df["loss"], label="Train Loss")
    plt.plot(history_df["epoch"], history_df["val_loss"], label="Val Loss")
    plt.legend()
    plt.title("Loss by Epoch")
    plt.xlabel("Epoch")

    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_training_curves.png", dpi=150)
    plt.show()
    plt.close()


def save_confusion_matrix(
    y_true: np.ndarray,
    y_pred: np.ndarray,
    split_name: str,
    class_names: list[str],
) -> None:
    cm = confusion_matrix(y_true, y_pred, labels=list(range(len(class_names))))
    cm_df = pd.DataFrame(cm, index=class_names, columns=class_names)
    cm_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_confusion_matrix_{split_name}.csv")

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names,
    )
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(f"Confusion Matrix - {split_name.upper()}")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_confusion_matrix_{split_name}.png", dpi=150)
    plt.show()
    plt.close()


def save_metric_bar_chart(summary: dict[str, float], split_name: str, focus_label: str) -> None:
    labels = ["Accuracy", "Macro Precision", "Macro Recall", "Macro F1", "Balanced Acc", f"{focus_label} Recall"]
    values = [
        summary["accuracy"],
        summary["precision_macro"],
        summary["recall_macro"],
        summary["f1_macro"],
        summary["balanced_accuracy"],
        summary["focus_recall"],
    ]

    plt.figure(figsize=(11, 5))
    bars = plt.bar(labels, values)
    plt.ylim(0, 1.05)
    plt.title(f"Evaluation Metrics - {split_name.upper()}")
    plt.ylabel("Score")
    for bar, value in zip(bars, values):
        plt.text(bar.get_x() + bar.get_width() / 2, value + 0.01, f"{value:.4f}", ha="center")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_metrics_{split_name}.png", dpi=150)
    plt.show()
    plt.close()


def save_class_report_heatmap(report_df: pd.DataFrame, split_name: str, class_names: list[str]) -> None:
    heatmap_df = report_df.loc[class_names, ["precision", "recall", "f1-score"]]
    plt.figure(figsize=(7, 3.8))
    sns.heatmap(heatmap_df, annot=True, fmt=".4f", cmap="YlGnBu", vmin=0, vmax=1)
    plt.title(f"Class Metrics Heatmap - {split_name.upper()}")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_class_metrics_heatmap_{split_name}.png", dpi=150)
    plt.show()
    plt.close()


def predict_generator_probabilities(model: tf.keras.Model, generator) -> np.ndarray:
    generator.reset()
    return model.predict(generator, verbose=1)


def evaluate_model(
    model: tf.keras.Model,
    generator,
    split_name: str,
    class_names: list[str],
    probabilities: np.ndarray | None = None,
) -> dict[str, object]:
    if probabilities is None:
        probabilities = predict_generator_probabilities(model, generator)

    y_pred = np.argmax(probabilities, axis=1)
    y_true = generator.classes

    report_dict = classification_report(
        y_true,
        y_pred,
        labels=list(range(len(class_names))),
        target_names=class_names,
        digits=4,
        output_dict=True,
        zero_division=0,
    )
    report_text = classification_report(
        y_true,
        y_pred,
        labels=list(range(len(class_names))),
        target_names=class_names,
        digits=4,
        zero_division=0,
    )
    report_df = pd.DataFrame(report_dict).transpose()

    focus_label = "bad" if "bad" in class_names else class_names[0]
    summary = {
        "split": split_name,
        "accuracy": accuracy_score(y_true, y_pred),
        "precision_macro": precision_score(y_true, y_pred, average="macro", zero_division=0),
        "recall_macro": recall_score(y_true, y_pred, average="macro", zero_division=0),
        "f1_macro": f1_score(y_true, y_pred, average="macro", zero_division=0),
        "precision_weighted": precision_score(y_true, y_pred, average="weighted", zero_division=0),
        "recall_weighted": recall_score(y_true, y_pred, average="weighted", zero_division=0),
        "f1_weighted": f1_score(y_true, y_pred, average="weighted", zero_division=0),
        "macro_f1": f1_score(y_true, y_pred, average="macro", zero_division=0),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "focus_label": focus_label,
        "focus_precision": report_dict[focus_label]["precision"],
        "focus_recall": report_dict[focus_label]["recall"],
        "focus_f1": report_dict[focus_label]["f1-score"],
    }

    for class_name in class_names:
        summary[f"{class_name}_precision"] = report_dict[class_name]["precision"]
        summary[f"{class_name}_recall"] = report_dict[class_name]["recall"]
        summary[f"{class_name}_f1"] = report_dict[class_name]["f1-score"]

    print(f"\n=== {split_name.upper()} REPORT ===")
    print(report_text)
    print("Summary:", summary)

    report_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_classification_report_{split_name}.csv")
    pd.DataFrame([summary]).to_csv(RESULTS_DIR / f"{MODEL_NAME}_summary_{split_name}.csv", index=False)

    save_confusion_matrix(y_true, y_pred, split_name, class_names)
    save_metric_bar_chart(summary, split_name, focus_label)
    save_class_report_heatmap(report_df, split_name, class_names)

    return {
        "summary": summary,
        "report_df": report_df,
        "y_true": y_true,
        "y_pred": y_pred,
        "probabilities": probabilities,
    }


def save_split_comparison(summaries: list[dict[str, float]]) -> None:
    summary_df = pd.DataFrame(summaries)
    summary_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_split_metrics.csv", index=False)
    display(summary_df)

    if summary_df.empty:
        return

    metric_columns = ["accuracy", "precision_macro", "recall_macro", "f1_macro", "focus_recall"]
    plot_df = summary_df[["split", *metric_columns]].set_index("split")
    ax = plot_df.plot(kind="bar", figsize=(11, 5))
    ax.set_ylim(0, 1.0)
    ax.set_ylabel("Score")
    ax.set_title("Split Comparison")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_split_comparison.png", dpi=150)
    plt.show()
    plt.close()


def save_test_predictions(
    generator,
    y_true: np.ndarray,
    y_pred: np.ndarray,
    probabilities: np.ndarray,
    class_names: list[str],
) -> None:
    data = {
        "filename": generator.filenames,
        "y_true": [class_names[index] for index in y_true],
        "y_pred": [class_names[index] for index in y_pred],
        "confidence": np.max(probabilities, axis=1),
    }
    for index, class_name in enumerate(class_names):
        data[f"probability_{class_name}"] = probabilities[:, index]
    pd.DataFrame(data).to_csv(RESULTS_DIR / f"{MODEL_NAME}_test_predictions.csv", index=False)


def preprocess_image_for_model(image_bgr: np.ndarray) -> np.ndarray:
    if image_bgr is None:
        raise ValueError("Cannot preprocess an empty image.")

    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    resized = cv2.resize(image_rgb, IMAGE_SIZE, interpolation=cv2.INTER_AREA).astype("float32")
    processed = resized / 255.0
    return np.expand_dims(processed, axis=0)


def predict_single_image(model: tf.keras.Model, img_path: Path, class_names: list[str]) -> tuple[str, float]:
    image = cv2.imread(str(img_path))
    processed = preprocess_image_for_model(image)
    probabilities = model.predict(processed, verbose=0)[0]
    predicted_index = int(np.argmax(probabilities))
    return class_names[predicted_index], float(probabilities[predicted_index])


def show_random_test_images(model: tf.keras.Model, test_dir: Path, class_names: list[str]) -> None:
    all_images: list[Path] = []
    for class_name in class_names:
        folder = test_dir / class_name
        if folder.exists():
            for image_path in folder.iterdir():
                if image_path.is_file() and image_path.suffix.lower() in IMAGE_SUFFIXES:
                    all_images.append(image_path)

    if not all_images:
        return

    sample_images = random.sample(all_images, min(4, len(all_images)))
    plt.figure(figsize=(8, 8))

    for index, image_path in enumerate(sample_images, start=1):
        image = cv2.imread(str(image_path))
        if image is None:
            continue

        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        pred_label, confidence = predict_single_image(model, image_path, class_names)
        true_label = image_path.parent.name

        plt.subplot(2, 2, index)
        plt.imshow(image_rgb)
        plt.title(f"Pred: {pred_label} ({confidence:.2%})\nTrue: {true_label}")
        plt.axis("off")

    plt.suptitle("Random Test Predictions")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_random_test_predictions.png", dpi=150)
    plt.show()
    plt.close()


def make_callbacks(checkpoint_path: Path) -> list[tf.keras.callbacks.Callback]:
    callbacks: list[tf.keras.callbacks.Callback] = [
        ModelCheckpoint(
            str(checkpoint_path),
            monitor="val_loss",
            save_best_only=True,
            mode="min",
            verbose=1,
        ),
        ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.2,
            patience=5,
            min_lr=1e-7,
            verbose=1,
        ),
    ]
    if USE_EARLY_STOPPING:
        callbacks.append(
            EarlyStopping(
                monitor="val_loss",
                patience=EARLY_STOPPING_PATIENCE,
                restore_best_weights=True,
                verbose=1,
            )
        )
    return callbacks


def main() -> None:
    configure_environment()

    if not PREPROCESSED_ROOT.exists():
        raise FileNotFoundError(f"Khong tim thay thu muc dataset: {PREPROCESSED_ROOT}")

    cleanup_old_outputs(RESULTS_DIR)
    train_gen, train_eval_gen, val_gen, test_gen = create_generators(PREPROCESSED_ROOT)
    class_names = list_class_names_from_indices(train_gen.class_indices)
    save_config(class_names)

    if len(class_names) < 2:
        raise ValueError(f"Can toi thieu 2 lop de huan luyen. Hien tai co {len(class_names)} lop: {class_names}")

    show_batch_samples(train_gen)
    class_weight = build_class_weight(train_gen.class_indices, train_eval_gen.classes)

    best_run_index = 1
    best_history_df = pd.DataFrame()
    best_val_loss = float("inf")
    best_val_macro_f1 = -1.0
    best_val_focus_recall = -1.0
    best_val_loss_list: list[float] = []
    run_rows: list[dict[str, float]] = []

    for run_index in range(RUNS):
        print(f"===== RUN {run_index + 1}/{RUNS} =====")
        model = create_model(num_classes=len(class_names))
        model.summary()

        checkpoint_path = RESULTS_DIR / f"{MODEL_NAME}_best_model_run{run_index + 1}.keras"
        callbacks = make_callbacks(checkpoint_path)

        history = model.fit(
            train_gen,
            epochs=EPOCHS,
            validation_data=val_gen,
            callbacks=callbacks,
            class_weight=class_weight,
            verbose=1,
        )

        history_df = pd.DataFrame(history.history)
        history_df.insert(0, "epoch", range(1, len(history_df) + 1))
        history_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_history_run{run_index + 1}.csv", index=False)

        run_best_val_loss = float(history_df["val_loss"].min())
        best_val_loss_list.append(run_best_val_loss)

        run_best_model = load_model(checkpoint_path)
        val_result = evaluate_model(run_best_model, val_gen, f"val_run{run_index + 1}", class_names)
        run_val_macro_f1 = float(val_result["summary"]["f1_macro"])
        run_val_focus_recall = float(val_result["summary"]["focus_recall"])

        run_rows.append({
            "run": run_index + 1,
            "best_val_loss": run_best_val_loss,
            "val_macro_f1": run_val_macro_f1,
            "val_focus_recall": run_val_focus_recall,
        })

        is_better_run = run_val_macro_f1 > best_val_macro_f1 or (
            np.isclose(run_val_macro_f1, best_val_macro_f1)
            and (
                run_val_focus_recall > best_val_focus_recall
                or (
                    np.isclose(run_val_focus_recall, best_val_focus_recall)
                    and run_best_val_loss < best_val_loss
                )
            )
        )
        if is_better_run:
            best_run_index = run_index + 1
            best_history_df = history_df
            best_val_loss = run_best_val_loss
            best_val_macro_f1 = run_val_macro_f1
            best_val_focus_recall = run_val_focus_recall

    pd.DataFrame(run_rows).to_csv(RESULTS_DIR / f"{MODEL_NAME}_run_summary.csv", index=False)

    print("=== OVERALL RESULT ===")
    print("Average best val loss:", np.mean(best_val_loss_list))
    print("Min best val loss:", min(best_val_loss_list))
    print("Best run:", best_run_index)
    print("Best validation macro F1:", best_val_macro_f1)
    print("Best validation focus recall:", best_val_focus_recall)

    if best_history_df.empty:
        return

    plot_history(best_history_df)
    best_history_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_epoch_metrics.csv", index=False)

    best_model_path = RESULTS_DIR / f"{MODEL_NAME}_best_model_run{best_run_index}.keras"
    best_model = load_model(best_model_path)
    summaries: list[dict[str, float]] = []

    train_result = evaluate_model(best_model, train_eval_gen, "train", class_names)
    summaries.append(train_result["summary"])

    val_result = evaluate_model(best_model, val_gen, "val", class_names)
    summaries.append(val_result["summary"])

    if test_gen is not None and test_gen.samples > 0:
        test_result = evaluate_model(best_model, test_gen, "test", class_names)
        summaries.append(test_result["summary"])
        save_test_predictions(
            test_gen,
            test_result["y_true"],
            test_result["y_pred"],
            test_result["probabilities"],
            class_names,
        )
        show_random_test_images(best_model, PREPROCESSED_ROOT / "test", class_names)

    save_split_comparison(summaries)
    print("Da luu toan bo ket qua vao:", RESULTS_DIR)


main()
```
