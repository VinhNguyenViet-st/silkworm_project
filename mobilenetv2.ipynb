```python
from __future__ import annotations

import json
import random
import shutil
import subprocess
import zipfile
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf
from IPython.display import display
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import ImageDataGenerator


MODEL_NAME = "mobilenetv2"
try:
    from google.colab import drive, files
    IN_COLAB = True
except Exception:
    drive = None
    files = None
    IN_COLAB = False

LOCAL_DATASET_DIR = Path("/content/dataset_classification")
LOCAL_DATASET_ZIP = Path("/content/dataset_classification.zip")
DRIVE_PROJECT_DIR = Path("/content/drive/MyDrive/BCKLTN")
DRIVE_DATASET_DIR = DRIVE_PROJECT_DIR / "dataset_classification"

RESULTS_ROOT = Path("/content/training_results") if IN_COLAB else Path(r"C:\Users\dungt\Downloads\BCKLTN\training_results")
RESULTS_DIR = RESULTS_ROOT / MODEL_NAME
ZIP_NAME = f"{MODEL_NAME}_training_results.zip"
ZIP_OUTPUT_PATH = Path("/content") / ZIP_NAME if IN_COLAB else RESULTS_ROOT / ZIP_NAME
DOWNLOAD_ZIP_AFTER_RUN = True

# Bat buoc dung GPU. Neu khong co GPU, notebook dung lai, khong train bang CPU.
REQUIRE_GPU = True
DEVICE_NAME = "/GPU:0"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32

# Tong 100 epoch: 50 epoch train head + 50 epoch fine-tune.
EPOCHS_HEAD = 50
EPOCHS_FINE_TUNE = 50
RUNS = 1

# De chay du 100 epoch, mac dinh tat EarlyStopping.
USE_EARLY_STOPPING = False
EARLY_STOPPING_PATIENCE = 10

DROPOUT_RATE = 0.3
LEARNING_RATE_HEAD = 1e-3
LEARNING_RATE_FINE_TUNE = 1e-5
FINE_TUNE_AT = 100
SEED = 42

EXPECTED_CLASS_ORDER = ("bad", "good", "shell")
IMAGE_SUFFIXES = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}


def configure_environment() -> None:
    global DEVICE_NAME

    random.seed(SEED)
    np.random.seed(SEED)
    tf.keras.utils.set_random_seed(SEED)

    print("TensorFlow version:", tf.__version__)
    print("TensorFlow build CUDA:", tf.test.is_built_with_cuda())

    if IN_COLAB:
        print("Dang chay tren Google Colab.")
        print("Kiem tra GPU bang nvidia-smi:")
        try:
            subprocess.run(["nvidia-smi"], check=False)
        except Exception as exc:
            print("Khong chay duoc lenh nvidia-smi:", exc)
    else:
        print("Dang chay ngoai Google Colab. Neu may ban khong cai TensorFlow GPU/CUDA thi se chi nhan CPU.")

    gpus = tf.config.list_physical_devices("GPU")
    print("Physical GPU devices:", gpus)

    if not gpus:
        DEVICE_NAME = "/CPU:0"
        message = (
            "KHONG TIM THAY GPU. Notebook da dung lai de tranh train bang CPU.\n"
            "Neu dung Google Colab: vao Runtime > Change runtime type > T4 GPU, "
            "sau do Runtime > Restart runtime va chay lai tu dau."
        )
        if REQUIRE_GPU:
            raise RuntimeError(message)
        print(message)
        return

    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError as exc:
        print("Khong the bat memory growth vi GPU da duoc khoi tao:", exc)

    DEVICE_NAME = "/GPU:0"
    print("XAC NHAN: Notebook se train bang GPU:", DEVICE_NAME)
    print("Logical devices:", tf.config.list_logical_devices())


def prepare_dataset_path() -> Path:
    """Tim dataset trong /content hoac Drive. Neu co zip trong /content thi tu giai nen."""
    if IN_COLAB and not LOCAL_DATASET_DIR.exists() and LOCAL_DATASET_ZIP.exists():
        print("Tim thay dataset_classification.zip, dang giai nen...")
        with zipfile.ZipFile(LOCAL_DATASET_ZIP, "r") as zip_ref:
            zip_ref.extractall("/content")
        print("Da giai nen xong.")

    if LOCAL_DATASET_DIR.exists():
        print("Su dung dataset local tren Colab:", LOCAL_DATASET_DIR)
        return LOCAL_DATASET_DIR

    # Chi mount Drive neu can tim dataset trong Drive.
    if IN_COLAB and drive is not None and not DRIVE_DATASET_DIR.exists():
        print("Khong thay dataset local. Dang mount Google Drive de tim dataset...")
        drive.mount("/content/drive")

    if DRIVE_DATASET_DIR.exists():
        print("Su dung dataset tren Google Drive:", DRIVE_DATASET_DIR)
        return DRIVE_DATASET_DIR

    raise FileNotFoundError(
        "Khong tim thay dataset_classification.\n"
        "Ban co the upload file /content/dataset_classification.zip, "
        "hoac giai nen thanh /content/dataset_classification, "
        "hoac dat tai /content/drive/MyDrive/BCKLTN/dataset_classification."
    )


def cleanup_old_outputs() -> None:
    if RESULTS_DIR.exists():
        shutil.rmtree(RESULTS_DIR)
    RESULTS_DIR.mkdir(parents=True, exist_ok=True)
    RESULTS_ROOT.mkdir(parents=True, exist_ok=True)


def list_class_names_from_indices(class_indices: dict[str, int]) -> list[str]:
    return [
        class_name
        for class_name, _ in sorted(class_indices.items(), key=lambda item: item[1])
    ]


def check_class_order(class_names: list[str]) -> None:
    expected = list(EXPECTED_CLASS_ORDER)
    if class_names != expected:
        print("CANH BAO: Thu tu lop hien tai:", class_names)
        print("Thu tu lop mong muon:", expected)
        print("Notebook dang ep thu tu lop theo EXPECTED_CLASS_ORDER neu du thu muc lop.")


def create_generators(data_root: Path):
    model_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

    common_kwargs = dict(
        target_size=IMAGE_SIZE,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        classes=list(EXPECTED_CLASS_ORDER),
    )

    train_gen = model_datagen.flow_from_directory(
        data_root / "train",
        shuffle=True,
        seed=SEED,
        **common_kwargs,
    )

    train_eval_gen = model_datagen.flow_from_directory(
        data_root / "train",
        shuffle=False,
        **common_kwargs,
    )

    val_gen = model_datagen.flow_from_directory(
        data_root / "val",
        shuffle=False,
        **common_kwargs,
    )

    test_dir = data_root / "test"
    test_gen = None
    if test_dir.exists():
        test_gen = model_datagen.flow_from_directory(
            test_dir,
            shuffle=False,
            **common_kwargs,
        )

    class_names = list_class_names_from_indices(train_gen.class_indices)
    check_class_order(class_names)
    print("Class mapping:", train_gen.class_indices)

    pd.DataFrame(
        [{"class_name": name, "class_index": index} for name, index in train_gen.class_indices.items()]
    ).sort_values("class_index").to_csv(RESULTS_DIR / f"{MODEL_NAME}_class_indices.csv", index=False)

    return train_gen, train_eval_gen, val_gen, test_gen


def build_class_weight(class_indices: dict[str, int], labels: np.ndarray) -> dict[int, float]:
    counts = np.bincount(labels.astype(int), minlength=len(class_indices))
    total = int(counts.sum())
    num_classes = len(class_indices)
    class_weight: dict[int, float] = {}

    class_count_by_name = {
        class_name: int(counts[index])
        for class_name, index in sorted(class_indices.items(), key=lambda item: item[1])
    }
    print("Class counts:", class_count_by_name)

    pd.DataFrame(
        [{"class_name": name, "count": count} for name, count in class_count_by_name.items()]
    ).to_csv(RESULTS_DIR / f"{MODEL_NAME}_class_counts.csv", index=False)

    for class_name, index in class_indices.items():
        class_count = int(counts[index])
        if class_count == 0:
            raise ValueError(f"Lop '{class_name}' khong co anh trong tap train.")
        class_weight[index] = total / (num_classes * class_count)

    print("Class weight:", class_weight)
    pd.DataFrame(
        [{"class_index": index, "class_weight": weight} for index, weight in class_weight.items()]
    ).to_csv(RESULTS_DIR / f"{MODEL_NAME}_class_weight.csv", index=False)
    return class_weight


def save_config(class_names: list[str], data_root: Path) -> None:
    config = {
        "model_name": MODEL_NAME,
        "dataset_path": str(data_root),
        "results_dir": str(RESULTS_DIR),
        "zip_output_path": str(ZIP_OUTPUT_PATH),
        "device_name": DEVICE_NAME,
        "image_size": IMAGE_SIZE,
        "batch_size": BATCH_SIZE,
        "epochs_head": EPOCHS_HEAD,
        "epochs_fine_tune": EPOCHS_FINE_TUNE,
        "total_epochs": EPOCHS_HEAD + EPOCHS_FINE_TUNE,
        "runs": RUNS,
        "class_names": class_names,
        "expected_class_order": EXPECTED_CLASS_ORDER,
        "learning_rate_head": LEARNING_RATE_HEAD,
        "learning_rate_fine_tune": LEARNING_RATE_FINE_TUNE,
        "fine_tune_at": FINE_TUNE_AT,
    }
    with open(RESULTS_DIR / f"{MODEL_NAME}_config.json", "w", encoding="utf-8") as f:
        json.dump(config, f, ensure_ascii=False, indent=2)


def show_batch_samples(generator) -> None:
    images, labels = next(generator)
    class_names = list_class_names_from_indices(generator.class_indices)

    # MobileNetV2 preprocess_input dua anh ve khoang [-1, 1], nen doi lai ve [0, 1] de hien thi.
    display_images = np.clip((images + 1.0) / 2.0, 0.0, 1.0)

    plt.figure(figsize=(12, 4))
    for i in range(min(6, len(display_images))):
        plt.subplot(2, 3, i + 1)
        plt.imshow(display_images[i])
        plt.title(class_names[int(np.argmax(labels[i]))])
        plt.axis("off")

    plt.suptitle("MobileNetV2 input preview")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_preprocessed_samples.png", dpi=150)
    plt.show()
    plt.close()
    generator.reset()


def create_model(num_classes: int) -> tuple[tf.keras.Model, tf.keras.Model]:
    base_model = MobileNetV2(
        input_shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3),
        include_top=False,
        weights="imagenet",
    )
    base_model.trainable = False

    inputs = layers.Input(shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3))
    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(DROPOUT_RATE)(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(DROPOUT_RATE)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = models.Model(inputs, outputs)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE_HEAD),
        loss="categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model, base_model


def unfreeze_for_fine_tuning(model: tf.keras.Model, base_model: tf.keras.Model) -> None:
    base_model.trainable = True

    for layer in base_model.layers[:FINE_TUNE_AT]:
        layer.trainable = False

    for layer in base_model.layers:
        if isinstance(layer, layers.BatchNormalization):
            layer.trainable = False

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE_FINE_TUNE),
        loss="categorical_crossentropy",
        metrics=["accuracy"],
    )


def history_to_dataframe(histories: list[tf.keras.callbacks.History]) -> pd.DataFrame:
    merged: dict[str, list[float]] = {}
    for history in histories:
        if history is None:
            continue
        for key, values in history.history.items():
            merged.setdefault(key, []).extend(values)

    history_df = pd.DataFrame(merged)
    history_df.insert(0, "epoch", range(1, len(history_df) + 1))
    return history_df


def plot_history(history_df: pd.DataFrame) -> None:
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history_df["epoch"], history_df["accuracy"], label="Train Accuracy")
    plt.plot(history_df["epoch"], history_df["val_accuracy"], label="Val Accuracy")
    plt.legend()
    plt.title("MobileNetV2 Accuracy by Epoch")
    plt.xlabel("Epoch")

    plt.subplot(1, 2, 2)
    plt.plot(history_df["epoch"], history_df["loss"], label="Train Loss")
    plt.plot(history_df["epoch"], history_df["val_loss"], label="Val Loss")
    plt.legend()
    plt.title("MobileNetV2 Loss by Epoch")
    plt.xlabel("Epoch")

    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_training_curves.png", dpi=150)
    plt.show()
    plt.close()


def save_confusion_matrix(
    y_true: np.ndarray,
    y_pred: np.ndarray,
    split_name: str,
    class_names: list[str],
) -> None:
    cm = confusion_matrix(y_true, y_pred, labels=list(range(len(class_names))))
    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names,
    )
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(f"MobileNetV2 Confusion Matrix - {split_name.upper()}")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_confusion_matrix_{split_name}.png", dpi=150)
    plt.show()
    plt.close()


def save_metric_bar_chart(summary: dict[str, float], split_name: str) -> None:
    labels = ["Accuracy", "Macro Precision", "Macro Recall", "Macro F1", "Balanced Acc"]
    values = [
        summary["accuracy"],
        summary["precision_macro"],
        summary["recall_macro"],
        summary["f1_macro"],
        summary["balanced_accuracy"],
    ]

    plt.figure(figsize=(10, 5))
    bars = plt.bar(labels, values)
    plt.ylim(0, 1.05)
    plt.title(f"MobileNetV2 Evaluation Metrics - {split_name.upper()}")
    plt.ylabel("Score")
    for bar, value in zip(bars, values):
        plt.text(bar.get_x() + bar.get_width() / 2, value + 0.01, f"{value:.4f}", ha="center")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_metrics_{split_name}.png", dpi=150)
    plt.show()
    plt.close()


def save_class_report_heatmap(report_df: pd.DataFrame, split_name: str, class_names: list[str]) -> None:
    heatmap_df = report_df.loc[class_names, ["precision", "recall", "f1-score"]]
    plt.figure(figsize=(6, 3.5))
    sns.heatmap(heatmap_df, annot=True, fmt=".4f", cmap="YlGnBu", vmin=0, vmax=1)
    plt.title(f"MobileNetV2 Class Metrics Heatmap - {split_name.upper()}")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_class_metrics_heatmap_{split_name}.png", dpi=150)
    plt.show()
    plt.close()


def predict_generator_probabilities(model: tf.keras.Model, generator) -> np.ndarray:
    generator.reset()
    return model.predict(generator, verbose=1)


def evaluate_model(
    model: tf.keras.Model,
    generator,
    split_name: str,
    class_names: list[str],
    probabilities: np.ndarray | None = None,
) -> dict[str, object]:
    if probabilities is None:
        probabilities = predict_generator_probabilities(model, generator)

    y_pred = np.argmax(probabilities, axis=1)
    y_true = generator.classes
    labels = list(range(len(class_names)))

    report_dict = classification_report(
        y_true,
        y_pred,
        labels=labels,
        target_names=class_names,
        digits=4,
        output_dict=True,
        zero_division=0,
    )
    report_text = classification_report(
        y_true,
        y_pred,
        labels=labels,
        target_names=class_names,
        digits=4,
        zero_division=0,
    )
    report_df = pd.DataFrame(report_dict).transpose()

    summary = {
        "split": split_name,
        "accuracy": accuracy_score(y_true, y_pred),
        "precision_macro": precision_score(y_true, y_pred, labels=labels, average="macro", zero_division=0),
        "recall_macro": recall_score(y_true, y_pred, labels=labels, average="macro", zero_division=0),
        "f1_macro": f1_score(y_true, y_pred, labels=labels, average="macro", zero_division=0),
        "precision_weighted": precision_score(y_true, y_pred, labels=labels, average="weighted", zero_division=0),
        "recall_weighted": recall_score(y_true, y_pred, labels=labels, average="weighted", zero_division=0),
        "f1_weighted": f1_score(y_true, y_pred, labels=labels, average="weighted", zero_division=0),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "focus_class": class_names[0],
        "focus_recall": report_dict[class_names[0]]["recall"],
    }

    for class_name in class_names:
        summary[f"{class_name}_precision"] = report_dict[class_name]["precision"]
        summary[f"{class_name}_recall"] = report_dict[class_name]["recall"]
        summary[f"{class_name}_f1"] = report_dict[class_name]["f1-score"]
        summary[f"predicted_{class_name}"] = int((y_pred == class_names.index(class_name)).sum())

    print(f"=== {split_name.upper()} CLASSIFICATION REPORT ===")
    print(report_text)
    print(
        "Macro Precision/Recall/F1: "
        f"{summary['precision_macro']:.4f} / {summary['recall_macro']:.4f} / {summary['f1_macro']:.4f}"
    )
    display(pd.DataFrame([summary]))
    display(report_df)

    pd.DataFrame([summary]).to_csv(RESULTS_DIR / f"{MODEL_NAME}_summary_{split_name}.csv", index=False)
    report_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_classification_report_{split_name}.csv")

    with open(RESULTS_DIR / f"{MODEL_NAME}_classification_report_{split_name}.txt", "w", encoding="utf-8") as f:
        f.write(report_text)

    save_confusion_matrix(y_true, y_pred, split_name, class_names)
    save_metric_bar_chart(summary, split_name)
    save_class_report_heatmap(report_df, split_name, class_names)

    return {
        "summary": summary,
        "report_df": report_df,
        "y_true": y_true,
        "y_pred": y_pred,
        "probabilities": probabilities,
    }


def save_split_comparison(summaries: list[dict[str, float]]) -> None:
    summary_df = pd.DataFrame(summaries)
    summary_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_split_metrics.csv", index=False)
    display(summary_df)

    if summary_df.empty:
        return

    plot_columns = ["accuracy", "precision_macro", "recall_macro", "f1_macro", "balanced_accuracy"]
    plot_df = summary_df[["split"] + plot_columns].set_index("split")
    ax = plot_df.plot(kind="bar", figsize=(11, 5))
    ax.set_ylim(0, 1.1)
    ax.set_ylabel("Score")
    ax.set_title("MobileNetV2 Split Comparison")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_split_comparison.png", dpi=150)
    plt.show()
    plt.close()


def save_test_predictions(
    generator,
    y_true: np.ndarray,
    y_pred: np.ndarray,
    probabilities: np.ndarray,
    class_names: list[str],
) -> None:
    data = {
        "filename": generator.filenames,
        "y_true_index": y_true,
        "y_true_label": [class_names[i] for i in y_true],
        "y_pred_index": y_pred,
        "y_pred_label": [class_names[i] for i in y_pred],
        "confidence": np.max(probabilities, axis=1),
    }
    for index, class_name in enumerate(class_names):
        data[f"prob_{class_name}"] = probabilities[:, index]

    pd.DataFrame(data).to_csv(RESULTS_DIR / f"{MODEL_NAME}_test_predictions.csv", index=False)


def preprocess_image_for_model(image_bgr: np.ndarray) -> np.ndarray:
    if image_bgr is None:
        raise ValueError("Cannot preprocess an empty image.")

    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    resized = cv2.resize(image_rgb, IMAGE_SIZE, interpolation=cv2.INTER_AREA).astype("float32")
    processed = preprocess_input(resized)
    return np.expand_dims(processed, axis=0)


def predict_single_image(model: tf.keras.Model, img_path: Path, class_names: list[str]) -> tuple[str, float]:
    image = cv2.imread(str(img_path))
    processed = preprocess_image_for_model(image)
    probabilities = model.predict(processed, verbose=0)[0]
    predicted_index = int(np.argmax(probabilities))
    return class_names[predicted_index], float(probabilities[predicted_index])


def show_random_test_images(model: tf.keras.Model, test_dir: Path, class_names: list[str]) -> None:
    all_images: list[Path] = []
    for class_name in class_names:
        folder = test_dir / class_name
        if folder.exists():
            for image_path in folder.iterdir():
                if image_path.is_file() and image_path.suffix.lower() in IMAGE_SUFFIXES:
                    all_images.append(image_path)

    if not all_images:
        return

    sample_images = random.sample(all_images, min(4, len(all_images)))
    plt.figure(figsize=(8, 8))

    for index, image_path in enumerate(sample_images, start=1):
        image = cv2.imread(str(image_path))
        if image is None:
            continue

        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        pred_label, confidence = predict_single_image(model, image_path, class_names)
        true_label = image_path.parent.name

        plt.subplot(2, 2, index)
        plt.imshow(image_rgb)
        plt.title(f"Pred: {pred_label} ({confidence:.2%})\nTrue: {true_label}")
        plt.axis("off")

    plt.suptitle("MobileNetV2 Random Test Predictions")
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f"{MODEL_NAME}_random_test_predictions.png", dpi=150)
    plt.show()
    plt.close()


def make_callbacks(checkpoint_path: Path) -> list[tf.keras.callbacks.Callback]:
    callbacks: list[tf.keras.callbacks.Callback] = [
        ModelCheckpoint(
            str(checkpoint_path),
            monitor="val_loss",
            save_best_only=True,
            mode="min",
            verbose=1,
        ),
        ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.2,
            patience=3,
            min_lr=1e-7,
            verbose=1,
        ),
    ]

    if USE_EARLY_STOPPING:
        callbacks.append(
            EarlyStopping(
                monitor="val_loss",
                patience=EARLY_STOPPING_PATIENCE,
                restore_best_weights=True,
                verbose=1,
            )
        )

    return callbacks


def zip_and_download_results() -> None:
    """Nen toan bo ket qua MobileNetV2 thanh 1 file .zip va tai xuong tren Colab."""
    if not RESULTS_DIR.exists():
        print("Chua co thu muc ket qua de nen:", RESULTS_DIR)
        return

    ZIP_OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
    if ZIP_OUTPUT_PATH.exists():
        ZIP_OUTPUT_PATH.unlink()

    archive_base = ZIP_OUTPUT_PATH.with_suffix("")
    shutil.make_archive(str(archive_base), "zip", root_dir=str(RESULTS_DIR))

    print("Da nen toan bo ket qua MobileNetV2 thanh file zip:", ZIP_OUTPUT_PATH)
    print("File zip gom model .keras, metric .csv/.txt, confusion matrix, bieu do va predictions neu co.")

    # Neu Drive da mount, luu them 1 ban zip vao Drive de tranh mat khi runtime ngat.
    drive_zip_dir = DRIVE_PROJECT_DIR / "training_results"
    try:
        if Path("/content/drive/MyDrive").exists():
            drive_zip_dir.mkdir(parents=True, exist_ok=True)
            drive_zip_path = drive_zip_dir / ZIP_NAME
            shutil.copy2(ZIP_OUTPUT_PATH, drive_zip_path)
            print("Da luu them file zip vao Google Drive:", drive_zip_path)
    except Exception as exc:
        print("Khong the copy zip vao Drive:", exc)

    if IN_COLAB and DOWNLOAD_ZIP_AFTER_RUN and files is not None:
        print("Dang mo hop thoai tai file zip ve may...")
        files.download(str(ZIP_OUTPUT_PATH))


def main() -> None:
    configure_environment()
    data_root = prepare_dataset_path()
    cleanup_old_outputs()

    train_gen, train_eval_gen, val_gen, test_gen = create_generators(data_root)
    class_names = list_class_names_from_indices(train_gen.class_indices)

    if len(class_names) != 3:
        raise ValueError(f"Notebook nay duoc cau hinh cho 3 lop bad/good/shell. Hien tai co {len(class_names)} lop: {class_names}")

    save_config(class_names, data_root)
    show_batch_samples(train_gen)
    class_weight = build_class_weight(train_gen.class_indices, train_eval_gen.classes)

    best_run_index = 1
    best_histories: list[tf.keras.callbacks.History] = []
    best_val_loss = float("inf")
    best_val_macro_f1 = -1.0
    best_val_focus_recall = -1.0
    best_val_loss_list: list[float] = []
    run_rows: list[dict[str, float]] = []

    for run_index in range(RUNS):
        print(f"===== RUN {run_index + 1}/{RUNS} =====")
        with tf.device(DEVICE_NAME):
            model, base_model = create_model(num_classes=len(class_names))
        model.summary()

        checkpoint_path = RESULTS_DIR / f"{MODEL_NAME}_best_model_run{run_index + 1}.keras"
        callbacks = make_callbacks(checkpoint_path)
        run_histories: list[tf.keras.callbacks.History] = []

        print("=== STAGE 1: Train classification head ===")
        with tf.device(DEVICE_NAME):
            history_head = model.fit(
                train_gen,
                epochs=EPOCHS_HEAD,
                validation_data=val_gen,
                callbacks=callbacks,
                class_weight=class_weight,
                verbose=1,
            )
        run_histories.append(history_head)

        if EPOCHS_FINE_TUNE > 0:
            print("=== STAGE 2: Fine-tune MobileNetV2 ===")
            with tf.device(DEVICE_NAME):
                unfreeze_for_fine_tuning(model, base_model)
                history_fine = model.fit(
                    train_gen,
                    epochs=EPOCHS_HEAD + EPOCHS_FINE_TUNE,
                    initial_epoch=len(history_head.history["loss"]),
                    validation_data=val_gen,
                    callbacks=callbacks,
                    class_weight=class_weight,
                    verbose=1,
                )
            run_histories.append(history_fine)

        run_history_df = history_to_dataframe(run_histories)
        run_history_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_history_run{run_index + 1}.csv", index=False)

        run_best_val_loss = float(run_history_df["val_loss"].min())
        best_val_loss_list.append(run_best_val_loss)

        with tf.device(DEVICE_NAME):
            run_best_model = load_model(checkpoint_path)
            val_result = evaluate_model(run_best_model, val_gen, f"val_run{run_index + 1}", class_names)

        run_val_macro_f1 = float(val_result["summary"]["f1_macro"])
        run_val_focus_recall = float(val_result["summary"]["focus_recall"])

        run_rows.append({
            "run": run_index + 1,
            "best_val_loss": run_best_val_loss,
            "val_macro_f1": run_val_macro_f1,
            "val_focus_recall": run_val_focus_recall,
        })

        is_better_run = run_val_macro_f1 > best_val_macro_f1 or (
            np.isclose(run_val_macro_f1, best_val_macro_f1)
            and (
                run_val_focus_recall > best_val_focus_recall
                or (
                    np.isclose(run_val_focus_recall, best_val_focus_recall)
                    and run_best_val_loss < best_val_loss
                )
            )
        )

        if is_better_run:
            best_run_index = run_index + 1
            best_histories = run_histories
            best_val_loss = run_best_val_loss
            best_val_macro_f1 = run_val_macro_f1
            best_val_focus_recall = run_val_focus_recall

    pd.DataFrame(run_rows).to_csv(RESULTS_DIR / f"{MODEL_NAME}_run_summary.csv", index=False)

    print("=== OVERALL RESULT ===")
    print("Average best val loss:", np.mean(best_val_loss_list))
    print("Min best val loss:", min(best_val_loss_list))
    print("Best run:", best_run_index)
    print("Best validation macro F1:", best_val_macro_f1)
    print("Best validation focus recall:", best_val_focus_recall)

    if not best_histories:
        return

    best_history_df = history_to_dataframe(best_histories)
    plot_history(best_history_df)
    best_history_df.to_csv(RESULTS_DIR / f"{MODEL_NAME}_epoch_metrics.csv", index=False)

    best_model_path = RESULTS_DIR / f"{MODEL_NAME}_best_model_run{best_run_index}.keras"
    with tf.device(DEVICE_NAME):
        best_model = load_model(best_model_path)

    summaries: list[dict[str, float]] = []

    with tf.device(DEVICE_NAME):
        train_result = evaluate_model(best_model, train_eval_gen, "train", class_names)
    summaries.append(train_result["summary"])

    with tf.device(DEVICE_NAME):
        val_result = evaluate_model(best_model, val_gen, "val", class_names)
    summaries.append(val_result["summary"])

    if test_gen is not None and test_gen.samples > 0:
        with tf.device(DEVICE_NAME):
            test_result = evaluate_model(best_model, test_gen, "test", class_names)
        summaries.append(test_result["summary"])
        save_test_predictions(
            test_gen,
            test_result["y_true"],
            test_result["y_pred"],
            test_result["probabilities"],
            class_names,
        )
        with tf.device(DEVICE_NAME):
            show_random_test_images(best_model, data_root / "test", class_names)

    save_split_comparison(summaries)
    print("Da luu toan bo ket qua vao:", RESULTS_DIR)

    # Nen va tai xuong toan bo ket qua giong YOLO.
    zip_and_download_results()


main()
```
